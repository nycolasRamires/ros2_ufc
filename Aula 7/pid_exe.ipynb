```python
import numpy as np
import matplotlib.pyplot as plt
import control
from ipywidgets import interact, FloatSlider
```

```python
# G(s) = wn^2 / (s^2 + 2*z*wn*s + wn^2)
z = 0.5
wn = 2.0
den = [1, 2*z*wn, wn**2]
G = control.tf(wn**2, den)

t = np.linspace(0, 40, 1000)

def plotar_resposta_cb(Kp, Ki, Kd):
    # C(s) = Kp + Ki/s + Kd*s
    C = control.tf([Kd, Kp, Ki], [1, 0])
    # T(s) = C(s)* G(s) / (1 + C(s) * G(s))
    T = control.feedback(C * G, 1)

    # resposta a uma entrada degrau unitário
    t_out, y_out = control.step_response(T, t)
    plt.figure(figsize=(10, 6))
    plt.plot(t_out, y_out, label='Resposta da Planta', color='blue', linewidth=2)

    plt.axhline(y=1.0, color='red', linestyle='--', label='Ref. (Degrau)')

    plt.title(f'resposta com Kp={Kp}, Ki={Ki}, Kd={Kd}')
    plt.xlabel('Tempo (s)')
    plt.ylabel('')
    plt.legend()
    plt.grid(True)
    plt.ylim(0, 1.5)
    plt.show()
```

```python
interact(plotar_resposta_cb,
         Kp=FloatSlider(value=2.0, min=0.0, max=10.0, step=0.1, description='Ganho Kp:'),
         Ki=FloatSlider(value=1.0, min=0.0, max=5.0, step=0.1, description='Ganho Ki:'),
         Kd=FloatSlider(value=0.5, min=0.0, max=5.0, step=0.1, description='Ganho Kd:'));
```
